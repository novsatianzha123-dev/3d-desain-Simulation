```python
# =========================================================
# CELL 1
# IMPORT LIBRARY
# =========================================================

import pandas as pd
```

```python
# =========================================================
# CELL 2
# LOAD FILE HASIL ANALISIS
# =========================================================

df = pd.read_csv("hasil_analisis_robot_bipedal.csv")

print(df.head())
```

```python
# =========================================================
# CELL 3
# MENAMPILKAN JUMLAH DATA
# =========================================================

print("Jumlah data:")
print(len(df))
```

```python
# =========================================================
# CELL 4
# MENAMPILKAN DATA STABIL
# =========================================================

stable_robot = df[df['Stability'] == 'STABLE']

print(stable_robot)
```

```python
# =========================================================
# CELL 5
# MENAMPILKAN DATA TIDAK STABIL
# =========================================================

not_stable_robot = df[df['Stability'] == 'NOT_STABLE']

print(not_stable_robot)
```

```python
# =========================================================
# CELL 6
# MENCARI TORQUE TERKECIL
# =========================================================

best_torque = df.loc[df['Torque_Nm'].idxmin()]

print("Torque terkecil:")
print(best_torque)
```

```python
# =========================================================
# CELL 7
# MENCARI TORQUE TERBESAR
# =========================================================

max_torque = df.loc[df['Torque_Nm'].idxmax()]

print("Torque terbesar:")
print(max_torque)
```

```python
# =========================================================
# CELL 8
# RATA-RATA TORQUE
# =========================================================

avg_torque = df['Torque_Nm'].mean()

print("Rata-rata torque:")
print(avg_torque)
```

```python
# =========================================================
# CELL 9
# RATA-RATA COM
# =========================================================

avg_com = df['COM_m'].mean()

print("Rata-rata COM:")
print(avg_com)
```

```python
# =========================================================
# CELL 10
# RATA-RATA INERSIA
# =========================================================

avg_inertia = df['Inertia'].mean()

print("Rata-rata inersia:")
print(avg_inertia)
```

```python
# =========================================================
# CELL 11
# WORKSPACE TERBESAR
# =========================================================

best_workspace = df.loc[df['Workspace_m'].idxmax()]

print("Workspace terbesar:")
print(best_workspace)
```

```python
# =========================================================
# CELL 12
# 10 DESAIN TERBAIK
# BERDASARKAN TORQUE TERKECIL
# =========================================================

top10 = df.sort_values(by='Torque_Nm').head(10)

print(top10)
```

```python
# =========================================================
# CELL 13
# FILTER DESAIN OPTIMAL
# =========================================================
#
# Kriteria:
# - torque kecil
# - COM kecil
# - inertia kecil
#
# =========================================================

optimal = df[
    (df['Torque_Nm'] < 0.9) &
    (df['COM_m'] < 0.05) &
    (df['Inertia'] < 0.01)
]

print(optimal)
```

```python
# =========================================================
# CELL 14
# MENYIMPAN HASIL OPTIMASI
# =========================================================

optimal.to_csv("desain_optimal_robot.csv", index=False)

print("File desain optimal berhasil disimpan")
```

```python
# =========================================================
# CELL 15
# KESIMPULAN ANALISIS
# =========================================================

print("==============================")
print("KESIMPULAN ANALISIS")
print("==============================")

print("Jumlah data:", len(df))

print("Torque minimum:",
      df['Torque_Nm'].min())

print("Torque maksimum:",
      df['Torque_Nm'].max())

print("Rata-rata torque:",
      df['Torque_Nm'].mean())

print("Rata-rata COM:",
      df['COM_m'].mean())

print("Rata-rata inersia:",
      df['Inertia'].mean())
```
